```python
n = int(input("Number of nodes"))
m = int(input("Number of edges"))
```

```python
am = [[-1 for i in range(n)] for j in range(m)]
for i in range(n):
    print(am[i])
```

```python
ind = [0 for i in range(n)]
outd = [0 for i in range(n)]
for i in range(m):
    x = input()
    s,d = int(x.split()[0]), int(x.split()[1])
    am[s][d] = 1
```

```python
for i in range(n):
    print(am[i])
```

```python
print(ind)
print(outd)
```

```python
st = []
visited = [0 for i in range(n)]
print(st)
print(visited)
```

```python
sn = int(input("Source node: "))
```

```python
st.append(sn)
```

```python
pr = []
visited[sn] = 1
while len(st) != 0:
    cn = st.pop(-1)
    pr.append(cn)
    for i in range(n):
        if am[cn][i] == 1 and visited[i] != 1:
            st.append(i)
            visited[i] = 1
```

```python
print(pr)
```
