```python
a = int(input())

for i in range(a):
    for j in range(a - i - 1):
        print(" ", end="")

    for k in range(i + 1):
        print("* ", end="")

    print()
```

```python
row = int(input())
col = int(input())
li = [[-1 for j in range(col)] for i in range(row)]
for i in range(row):
    print(li[i])
```

```python
lii = [10,22,7,8,6,7,8,7,5,74,7,7]
n = len(lii)
for i in range(n-1):
    for j in range(n-i-1):
        if lii[j] > lii[j+1]:
            lii[j],lii[j+1] = lii[j+1],lii[j]


print(lii)
            
```
