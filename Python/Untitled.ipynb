```python
n = int(input("num of nodes"))
m = int(input("num of edges"))

              
```

```python
am = [[-1 for i in range(n)] for j in range(n)] 
for i in range(n):
    print(am[i])
```

```python
for i in range(m):
    x = input()
    s, d = int(x.split()[0]),int(x.split()[1])
    am[s][d] = 1
```

```python
for i in range(n):
    out = 0
    indeg = 0
    for j in range(n):
        out += am[i][j]
        indeg += am[j][i]
  print(i,"in degree",indeg,"out degree",out)
```
