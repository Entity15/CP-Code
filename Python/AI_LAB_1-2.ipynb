** 2D Array

```python
r = int(input())
c = int(input())

ar = [[0 for i in range(c)] for j in range(r)] # col for Each Row, run Row times

print(ar)
```

```python
ar[0][0] = 15
ar[0][1] = 8
ar[0][2] = 3
ar[1][0] = 10
ar[1][1] = 5
ar[1][2] = 4
ar[2][0] = 3
ar[2][1] = 2
ar[2][2] = 1
for i in range(r):
  print(ar[i])
```

```python
for i in range (r - 1):
  for j in range(r - i - 1):
    if ar[j][1] > ar[j + 1][1]:  #only adding [1] for sorting that col wise
      ar[j], ar[j + 1] = ar[j + 1], ar[j]
  print()

for i in range(r):
  print(ar[i])




# Homework sort by adding all row elements, if sum of row elements smaller put it first, larger last
```

```python
for i in range (r - 1):
  for j in range(r - i - 1):
    if sum(ar[j]) > sum(ar[j + 1]):  #only adding [1] for sorting that col wise
      ar[j], ar[j + 1] = ar[j + 1], ar[j]
  print()

for i in range(r):
  print(ar[i])
```
